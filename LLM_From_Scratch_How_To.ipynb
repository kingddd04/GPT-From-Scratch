### So you want to build an LLM From Scratch?

LLM are becoming more and more a part of our daily life, we love to make conjectures but often we have no idea how they work, the porpuse of this project is to show trough some simple python how this new tecnology works.  
  
With LLM we're referring to large language models, a special type of ai, capable of learning and generating high-quality textual content.
  
Let's start from the foundamentals, what is the GPT architecture? it stands for :  
- **Generative** -> Capable of generating new contents
- **Pre Trained** -> Trained before usage
- **Transformer** -> The "Brain" of the LLM a mathematical operation capable of extracting datas from the text

Building an llm from scratch is not something that is usually in range of the average python newbie, but in fact is just a big topic and like all big complex topics to be solved needs to be divided in smaller steps, that can be completed sequentially.
